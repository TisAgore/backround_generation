```python
# Короч качай мою зипку, там лежит TextGenerator исправленный
# Ес чо зайди в дискорд, я пока что там сижу

import unicodedata
from TextGenerator.text_generator import TextGenerator
import tqdm
import time

g = TextGenerator()
g.max_retries = 5

total = 20_000
amount = 100
sentences = []
for i in tqdm.tqdm(range(0, total, amount)):
    try:
        s = g.generate_text_chunks(amount=amount, max_words=7,)
    except RuntimeError as e:
        print(f"Exception: {e}, i: {i}")
        s = []
        time.sleep(30)

    sentences.extend(s)

```

```python
sentences = [sentence.replace("'", "").replace("’", "").replace("`", "").replace("/", "").
             replace("\\", "").replace('«', '').replace('»', '') for sentence in sentences]

def cleaner(sentences):
    temp = []
    for sentence in sentences:
        normalized = unicodedata.normalize('NFD', sentence)
        no_diacritics = ''.join(c for c in normalized if unicodedata.category(c) != 'Mn')
        no_diacritics = no_diacritics + '\n'
        temp.append(no_diacritics)
    return temp

normalized_sentences = cleaner(sentences)
len(normalized_sentences)
with open('test.txt', 'w', encoding='utf-8') as file:
    file.writelines(normalized_sentences)
```
